```python
import sqlite3
```

```python
connection = sqlite3.connect('movies.db')
```

```python
cursor = connection.cursor()
```

```python
#Kac adet tablomuz db'de ve isimleri neler ?
cursor.execute("SELECT name FROM sqlite_master WHERE type ='table'")
print(cursor.fetchall())
```

```python
#movies tablosu:
cursor.execute("SELECT sql FROM sqlite_master WHERE name ='movies'")
print(cursor.fetchall())
```

```python
#stars tablosu:
cursor.execute("SELECT sql FROM sqlite_master WHERE name ='stars'")
print(cursor.fetchall())
```

```python
#movies tablosunda kac tane film var
cursor.execute("SELECT COUNT(*) FROM movies")
print(f'Movies tabloumuzda {cursor.fetchall()} film var.')
```

```python
#people tablosunda kac tane insan
cursor.execute("SELECT COUNT(*) FROM people")
print(f'People tabloumuzda {cursor.fetchall()} adet insan var.')
```

```python
cursor.execute("SELECT title, year FROM movies LIMIT 10")
print(cursor.fetchall())
```

```python
cursor.execute("SELECT name FROM people LIMIT 10")
print(cursor.fetchall())
```

```python
#2000 yilindaki ilk 10 filmi var.
cursor.execute("SELECT title FROM movies WHERE year == 2000 LIMIT 10")
print(cursor.fetchall())
```
