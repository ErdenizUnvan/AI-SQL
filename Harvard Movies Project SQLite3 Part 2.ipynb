```python
import sqlite3
```

```python
connection = sqlite3.connect('movies.db')
```

```python
cursor = connection.cursor()
```

```python
#01) Show 20 oldest movies
#movies tablosunu kullanicaz
#id, title ,year
cursor.execute("SELECT title, year FROM  movies ORDER BY year LIMIT 20")
print(cursor.fetchall())
```

```python
#Show 20 youngest movies
#movies tablosunu kullanicaz
#id, title ,year
cursor.execute("SELECT title, year FROM  movies ORDER BY year DESC LIMIT 40")
print(cursor.fetchall())
```

```python
#en yasli 20 kisi kim
#people
#id, name, birth
cursor.execute("SELECT name, birth FROM people WHERE birth > 0 ORDER BY birth LIMIT 20")
print(cursor.fetchall())
```

```python
#en genc 20 kisi kim
#people
#id, name, birth
cursor.execute("SELECT name, birth FROM people WHERE birth > 0 ORDER BY birth DESC LIMIT 20")
print(cursor.fetchall())
```

```python
#Show Al Pacino and Robert De Niro birthday
#people
#id, name, birth
cursor.execute("SELECT name, birth FROM  people WHERE name =='Al Pacino' OR name =='Robert De Niro'")
print(cursor.fetchall())
```

```python
#Show all Matrix movies with chronological order
#movies
#id, title, year
cursor.execute("SELECT title, year FROM movies WHERE UPPER(title) LIKE '%MATRIX%' ORDER BY year")
print(cursor.fetchall())
```
